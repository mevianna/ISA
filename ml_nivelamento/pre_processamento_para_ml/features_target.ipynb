# Aula: Features e Target em Aprendizado de Máquina

## Introdução

Nesta aula, será apresentado o conceito de *features* e *target*, elementos fundamentais para a construção de qualquer modelo de aprendizado de máquina supervisionado. Compreender adequadamente essa distinção constitui pré-requisito indispensável para a modelagem de problemas de regressão e classificação.


## 1. O que são Features

Denomina-se *feature* (em português, "atributo" ou "variável explicativa") cada uma das colunas de um conjunto de dados que é utilizada como entrada para um modelo preditivo. As features representam as características observadas de um fenômeno, a partir das quais se pretende explicar ou prever um determinado resultado.

**Exemplo:** em um conjunto de dados sobre imóveis, poderiam ser consideradas features a área construída, o número de quartos, a localização e a idade do imóvel.

```python
import pandas as pd

dados = pd.DataFrame({
    "area_m2": [50, 80, 120, 200],
    "quartos": [1, 2, 3, 4],
    "idade_anos": [10, 5, 2, 20],
    "preco": [150000, 250000, 400000, 500000]
})

dados.head()


#Explicação do código:

# - `import pandas as pd`: importa a biblioteca pandas, responsável pela manipulação de dados tabulares, e a associa ao apelido `pd`,
#convenção amplamente adotada pela comunidade.

# - `pd.DataFrame({...})`: cria uma tabela de dados (DataFrame) a partir de um dicionário Python, em que cada chave se torna o nome 
#de uma coluna e cada lista associada se torna os valores dessa coluna.

# - `dados = ...`: armazena o DataFrame criado na variável `dados`.

# - `dados.head()`: exibe as primeiras linhas do DataFrame, permitindo inspecionar visualmente sua estrutura.

#Nesse exemplo, `area_m2`, `quartos` e `idade_anos` constituem as features do problema, pois representam informações de entrada.
```

## 2. O que é Target

Denomina-se *target* (em português, "alvo" ou "variável resposta") a coluna do conjunto de dados que representa o resultado que se deseja prever ou explicar. Trata-se, portanto, da variável de saída do modelo.

**Exemplo:** no conjunto de dados apresentado anteriormente, o `preco` do imóvel constitui o target, uma vez que é o valor que se pretende estimar a partir das demais colunas.

```python
import pandas as pd

dados = pd.DataFrame({
    "area_m2": [50, 80, 120, 200],
    "quartos": [1, 2, 3, 4],
    "idade_anos": [10, 5, 2, 20],
    "preco": [150000, 250000, 400000, 500000]
})

dados.head()

target = dados["preco"]

target

#Explicação do código:

#dados["preco"]: seleciona apenas a coluna preco do DataFrame, utilizando a notação de colchetes.
#target = ...: armazena essa coluna selecionada na variável target, por convenção de nomenclatura.
#target: exibe o conteúdo da variável, permitindo verificar que se trata de uma série (Series) de valores numéricos.
```

## 3. Variáveis Independentes e Dependentes


As **variáveis independentes** são aquelas cujo valor não depende de nenhuma outra variável do modelo; são fornecidas como entrada e presumidas como causas ou fatores explicativos.
A **variável dependente** é aquela cujo valor se supõe ser influenciado pelas variáveis independentes; é o efeito ou resultado que o modelo busca explicar.

Assim, ao afirmar que o preço de um imóvel "depende" da sua área e do número de quartos, estabelece-se uma relação de dependência estatística entre essas grandezas — daí a origem da terminologia.

| Denominação em Estatística | Denominação em Machine Learning | Papel no modelo |
|---|---|---|
| Variável independente | Feature | Entrada (causa) |
| Variável dependente | Target | Saída (efeito) |




## 4. Separação entre X e y

Por convenção consolidada na literatura de aprendizado de máquina, as features são representadas pela letra maiúscula **X** (por se tratar, em geral,
de uma matriz com múltiplas colunas), enquanto o target é representado pela letra minúscula **y** (por se tratar, em geral, de um vetor unidimensional).
Essa separação é etapa obrigatória antes do treinamento de qualquer modelo supervisionado, pois os algoritmos exigem que entradas e saídas sejam
fornecidas de forma distinta.

```python
X = dados.drop(columns=["preco"])
y = dados["preco"]

print(X.shape)
print(y.shape)

#Explicação do código acima:

# dados.drop(columns=["preco"])`: remove a coluna `preco` do DataFrame original, retornando um novo DataFrame apenas com as colunas restantes. 
# O parâmetro `columns=[...]` indica quais colunas serão excluídas.
# X = ...`: por convenção, a matriz de features é armazenada na variável de nome `X` (maiúscula).
# dados["preco"]`: seleciona novamente a coluna `preco`, que representa o target.
# y = ...`: por convenção, o vetor do target é armazenado na variável de nome `y` (minúscula).
# X.shape`: retorna uma tupla (número de linhas, número de colunas), indicando as dimensões da matriz de features.
# y.shape`: retorna o número de elementos do vetor target, confirmando tratar-se de uma única dimensão.


#Após essa separação, `X` e `y` estão prontos para serem utilizados, por exemplo, na divisão entre conjuntos de treino e teste,
#e posteriormente no treinamento de um modelo. 
```

## Síntese

| Conceito | Definição |
|---|---|
| Feature | Variável de entrada, utilizada para explicar ou prever o target |
| Target | Variável de saída, aquilo que se deseja prever |
| Variável independente | Sinônimo estatístico de feature |
| Variável dependente | Sinônimo estatístico de target |
| X | Convenção de nomenclatura para a matriz de features |
| y | Convenção de nomenclatura para o vetor do target |
