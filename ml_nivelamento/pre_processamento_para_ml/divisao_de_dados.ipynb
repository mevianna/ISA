# Divisão de Dados

(Data Splitting) A divisão dos dados é o processo de pegar o seu conjunto de dados original (dataset) e dividi-lo em duas ou três partes distintas.
A abordagem mais comum é dividir o dataset em: 

**Dados de Treino**: A maior parte dos dados, usada para o algoritmo aprender os padrões.

**Dados de Teste**: Uma parte menor, mantida em segredo durante o treinamento, usada apenas para a avaliação final do modelo.(Opcional)

**Dados de Validação**: Uma terceira parte usada para ajustar os hiperparâmetros (configurações) do modelo antes do teste final.

# Dados de Treino e Teste
Esses dois blocos possuem funções completamente diferentes no ciclo de desenvolvimento:

**Dados de Treino (Training Set)**: É o livro de estudos do modelo. O algoritmo analisa esses dados repetidamente para encontrar regras, pesos e correlações entre as características (features) e a resposta esperada (target).

**Dados de Teste (Testing Set)**: É a prova final do modelo. Ele serve para simular o comportamento do modelo em produção. Como o modelo nunca viu esses dados antes, o desempenho aqui dita o quão confiável o sistema realmente é.

# Por que separar os dados?

Separamos os dados principalmente para evitar o **Overfitting** (Sobreajuste).Se você treinar e testar o seu modelo usando os mesmos dados, ele terá uma precisão excelente (perto de 100%), mas isso acontece porque ele simplesmente memorizou o histórico. Ao receber novos dados do mundo real, ele falhará. Separar os dados permite medir a capacidade de generalização do modelo, ou seja, o quão bem ele aprendeu as regras gerais em vez de decorar exemplos específicos.

# train_test_split

É uma função da biblioteca Scikit-Learn (uma das mais populares do Python para Machine Learning).Essa função automatiza o processo de divisão de forma aleatória. Em vez de você selecionar manualmente quais linhas vão para treino e quais vão para teste, a função embaralha os dados e faz a separação de forma rápida e segura com apenas uma linha de código.

# test_size

É um parâmetro da função train_test_split que define a porcentagem (ou o número absoluto) de dados que será reservada para o conjunto de teste.Ele aceita valores decimais entre 0.0 e 1.0.Por exemplo, definir test_size=0.2 significa que 20% dos seus dados serão guardados para teste, enquanto os 80% restantes serão usados para o treino.Se não for especificado, o padrão geralmente adota o valor de 0.25 (25% para teste).

# random_state

É o parâmetro responsável por controlar a reprodutibilidade da divisão aleatória. Como o algoritmo precisa embaralhar os dados antes de dividir, cada vez que você rodar o código, a separação será diferente. Isso pode fazer com que a precisão do seu modelo mude a cada execução.Ao definir o random_state com qualquer número inteiro (ex: random_state=42), você define uma "semente" para o gerador de números aleatórios.Isso garante que o código sempre dividirá os dados exatamente da mesma forma toda vez que for executado, permitindo que outros programadores repliquem exatamente os seus resultados.

```python
# 1. Importar as bibliotecas necessárias
import numpy as np
from sklearn.model_selection import train_test_split

# 2. Criar dados fictícios (10 casas: Tamanho vs Preço)
# X representa as características (features) e y o objetivo (target)
X = np.array([[50], [60], [70], [80], [90], [100], [110], [120], [130], [140]])
y = np.array([150, 180, 210, 240, 270, 300, 330, 360, 390, 420])

# 3. Aplicar o train_test_split
# Reservamos 30% dos dados para teste (test_size=0.3)
# Fixamos a semente aleatória para sempre ter o mesmo resultado (random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 4. Exibir os resultados da divisão
print("- CONJUNTO DE TREINO (70%) -")
print("X_train (Tamanhos):\n", X_train)
print("y_train (Preços):", y_train)

print("\n- CONJUNTO DE TESTE (30%) -")
print("X_test (Tamanhos):\n", X_test)
print("y_test (Preços):", y_test)
```

| Conceito | Definição |
|---|---|
| Dados de treino | Subconjunto utilizado para o modelo aprender os padrões |
| Dados de teste | Subconjunto utilizado para avaliar o desempenho do modelo |
| train_test_split | Função do scikit-learn que realiza a divisão dos dados |
| test_size | Parâmetro que define a proporção de dados destinada ao teste |
| random_state | Parâmetro que fixa a aleatoriedade, garantindo reprodutibilidade |
